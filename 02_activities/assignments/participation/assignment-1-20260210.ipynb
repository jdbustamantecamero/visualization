```python
# Import Libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Generate reproducible sample data
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

# Create a figure with a single subplot
fig, ax = plt.subplots(figsize=(5,3))

#ax.scatter(x,y)
#ax.bar(x,y)
#ax.barh(x,y)
#ax.plot(x,y)

# Plot a histogram
ax.hist(y)
```

```python
# Define font styles
font1 = {'family':'sans-serif', 'color':'blue', 'size':20}
font2 = {'family':'monospace', 'color':'green', 'size':14}

# Create a line plot with custom-styled title and axis labels
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title('Total Growth Over Time', fontdict = font1, loc = 'left')
ax.set_ylabel('Total Groth', fontdict = font2)
ax.set_xlabel('Years Since Start', fontdict = font2)
fig.tight_layout()
```

```python
# Create a scatter plot with custom-styled markers
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y,marker='*',color='indigo')
```

```python
# Create a line plot with custom-styled lines and markers
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y,marker='*',color='indigo',linestyle='--',linewidth=2)
```

```python
# Create a line plot with custom-styled lines, markers, and grid
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y,
        marker='*',
        markersize=12,
        color='#7425b9',
        linestyle='--',
        linewidth=2,
        markeredgecolor='#fa9659',
        markerfacecolor='#000000')
ax.grid(linestyle='dotted',
        linewidth=1,
        color='#fa9659'
        )
```
