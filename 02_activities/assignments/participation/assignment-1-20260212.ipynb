```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label='Person 1')
ax.plot(x,y2, label='Person 2')
ax.legend(loc='lower right',
          frameon=True,
          fontsize=12,
          ncol=2,
          shadow=True)
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label='Person 1')
ax.plot(x,y2, label='Person 2')
ax.legend(loc='upper left',
          bbox_to_anchor=(1,1))
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label='Person 1')
ax.scatter(x,y2, label='Person 2')
ax.legend(loc='lower right')
ax.text(10,95,"This value is important!",
        ha='center',
        color='red',
        size=20)
fig.show()
```

```python
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Ussop", "Sanji"])
y2 = np.array([110,180,240,99,220])
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2,
                               nrows=1,
                               figsize=(9,3))

font1 = {'family':'serif', 'size':10}
font2 = {'family':'serif', 'size':14}

ax1.scatter(x1,y1, marker = 'o', color='red')
ax2.bar(x2,y2, color='darkgreen')

ax1.set_title('Important Scatterplot', fontdict=font2)
ax1.set_ylabel('Measure', fontdict=font1)
ax1.set_xlabel('Time', fontdict=font1)

ax2.set_title('Important Bar', fontdict=font1)
ax2.set_ylabel('Measure', fontdict=font2)
ax2.set_xlabel('Character', fontdict=font2)
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                   ['ax2', 'ax3']],
                                   figsize=(7,4),
                                   layout='constrained')

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)

someaxes['ax1'].set_xlabel('A Big Label', fontsize=18)
someaxes['ax2'].set_xlabel('Another Label', fontsize=18)
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)

plt.show()
```

```python
x = np.array(["Luffy", "Zoro", "Nami", "Ussop", "Sanji"])
y1 = np.array([110,180,240,99,220])
y2 = np.array([170,100,90,120,50])

fig, ax = plt.subplots(figsize=(7,3))
ax.bar(x,y1, color='indigo')
ax.plot(x,y2, color='red')
```

```python
y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x,y2, color='red')

ax.errorbar(x,y2,yerr=y2_sd, fmt='none',ecolor='indigo',elinewidth=4,capsize=6,capthick=4)
```
